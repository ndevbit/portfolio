Popcorn Hack 2

```python
def is_priority(incident_type):
    return (incident_type == "Medical Emergency")

def process_report(location, incident_type):
    is_priority_incident = is_priority(incident_type)
    if is_priority_incident:
        print(f"A {incident_type} is a priority, one has happened at {location}")
    else:
        print(f"A {incident_type} happened at {location}")
        
process_report("Area 1", "Rebellion")
process_report("Area 2", "Medical Emergency")
```

```python
def check_priority(incident_type):
    return (incident_type == "Forest Fire")
    
def dispatch_response(location, incident_type):
    is_priority = check_priority(incident_type)
    if (is_priority):
        print(f"""An incident with priority has happened at {location}, 
that incident is a {incident_type}""")
    else:
        print(f"A {incident_type} has happened at {location}")
        
dispatch_response("Keyboard City", "Forest Fire")
dispatch_response("Cool lake town", "Train arrival")
dispatch_response("Epic area", "Theme park opening")
```
