```python
raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

stripped = raw_transmission.strip()
fields = stripped.split(",")

print(fields)

date = fields[0]
agent = fields[1]
message_code = fields[2]
confidence_str = fields[3]

print(f"Date: {date} | Agent: {agent} | Message Code: {message_code} | Confidence: {confidence_str}")

year_check = date[0:4]
print("Year check:", year_check)

message = message_code.replace("-", " ")

found_dawn = "dawn" in message
print("Found Dawn:", found_dawn)

markdown_row = f"{date} | {agent} | {message} | {confidence_str}"
print("Markdown row:", markdown_row)

json_line = "{" + f"""
    \"date\": \"{date}\"
    \"agent\": \"{agent}\"
    \"message\": \"{message}\"
    \"confidence\": \"{confidence_str}\"
    \"contains_dawn\" \"{str(found_dawn).lower()}\"
""" + "}"
print("JSON:", json_line)
```
