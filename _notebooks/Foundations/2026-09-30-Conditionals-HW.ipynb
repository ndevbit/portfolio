```python
# CODE RUNNER

message = "gift card"

risk_score = 0
amount_requested = 0
known_contact = False

# There is no need for an else statement in the Find gift cards,
# Amount requested, and Known contact sections of the code
# as there is nothing that needs to be done should
# the condition not be met

# Find gift cards
found_gift_cards = False

target_string = "gift card"
for i in range(len(message)):
    if message[i:i+len(target_string)].lower() == target_string:
        risk_score += 4
        found_gift_cards = True
        
# Amount requested       
over_100 = False

if amount_requested > 100:
    risk_score += 3
    over_100 = True

# Known contact
if known_contact != True:
    risk_score += 2

# Print results
print("Input:")
print(f"Message: {message}")
print(f"Amount requested: {amount_requested}")
print(f"Known Contact?: {known_contact}")
print("""
Assessment:""")
print(f"Risk Score: {risk_score}")
print(f"Amount reqeusted over 100?: {over_100}")
print(f"Message said something about gift cards?: {found_gift_cards}")
if found_gift_cards: print("Gift card mentioned in the message.")

print("""
Recommendation:""")
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("This is definitely a scam, do NOT engage")
elif risk_score >= 7:
    print("This might be a scam, only check it if you can be sure it's not a scam")
elif risk_score >= 0:
    print("This is probably not a scam, it should be fine")
else:
    print("This is not a scam, period")
```
