## Popcorn Hack

```python
# CODE_RUNNER:

# Theme: The SFSRC team scans wildfire alert levels from different zones.
# Goal: skip safe zones, count response zones, total their alert scores, and stop at the first critical zone.

alerts = [1, 4, 2, 6, 3, 9, 5]

# TODO: Pick the cutoff for safe alerts. Alerts below this number should be skipped.
safe_cutoff = 6

# TODO: Pick the cutoff for critical alerts. The loop should stop at this number or higher.
critical_cutoff = 8

# Counters and totals start at 0 because no alerts have been processed yet.
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

# TODO: Loop through the alert list.
for alert in alerts:
    # TODO: If the alert is safe, count it, print a message, and skip to the next alert.
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    # If code reaches here, the alert needs action.
    # TODO: Count this action alert and add its score to the running total.
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    # TODO: If this alert is critical, mark it and stop the scan.
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

# Final report: these lines should explain what happened during the scan.
print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

## Homework Hack

```python
# CODE_RUNNER:

# Input list
Volumes = [231, 334, 325, 213, 241, 716, 
           162, 323, 353, 315, 544]

# Thresholds
Cutoff_Low = 700
Cutoff_High = 500

TooLoudThreshold = 1000

# Summary variables
Modified_Sample_Count = 0
Samples_TooSmall_Count = 0
Samples_TooBig_Count = 0
TooLoudFound = False

# Find something too loud and report it
for i in range(len(Volumes)):
    if Volumes[i] < Cutoff_High:
        continue # Too quiet? just ignore it
    if Volumes[i] > TooLoudThreshold:
        TooLoudFound = True # Found something too loud, stop
        break

Result = []

# Copy Volumes into Result
for i in range(len(Volumes)):
    Result.append(Volumes[i])

# Clamp Volumes to loud
for i in range(len(Result)):
    if Result[i] > Cutoff_Low:
        Result[i] = Cutoff_Low
        Modified_Sample_Count += 1
        Samples_TooBig_Count += 1
        
# Clamp Volumes too quiet
for i in range(len(Result)):
    if Result[i] < Cutoff_High:
        Result[i] = Cutoff_High
        Modified_Sample_Count += 1
        Samples_TooSmall_Count += 1
        
# Print results
print(Volumes)
        
print(Result)
print(f"Modified_Sample_Count: {Modified_Sample_Count}")
print(f"Samples_TooSmall_Count: {Samples_TooSmall_Count}")
print(f"Samples_TooBig_Count: {Samples_TooBig_Count}")
print("---")
print(f"TooLoudThreshold: {TooLoudThreshold}")
print(f"TooLoudFound: {TooLoudFound}")
```
