B. Popcorn Hack "Expand the catalog"

````python
import math

n = 16
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)

# Prediction, given the worst case:
# Linear search will need 8 checks
# Binary search will need 3 checks (log2(8))

# Prediction, given n changes from 8 to 16:
# Linear search doubles
# Binary search goes up by 1

# I recommend binary search, because of its efficiency when compared to linear search
````

C. Popcorn Hack "Compare growth"

````python
# Prediction, given 10 books:
# Visits: 10
# Pairs: 100

n = 10
visits = n
pairs = n * n
print(n)
print(visits)
print(pairs)

# Doubling n doubles the visits because the visits scale with O(n) complexity, because
# the librarian only needs to visit each book one time, so 10 books means 10 total visits
# while pairs scale with O(n^2) time complexity (each book compare involves 10 checks
# ammssing to 100 checks in total (10 books * 10 checks/book))
````
