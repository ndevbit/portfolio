```python
# CODE_RUNNER:

import math
import random

# Read the API before calling a procedure
print("random.randint:", random.randint.__doc__)
print("math.ceil:", math.ceil.__doc__)
print(dir(math))
print(dir(random))

points_per_game = [24, 18, 31, 27, 22]

# random.choice picks one item out of a list
print("Random game to review:", random.choice(points_per_game), "points")

# random.randint(a, b) returns a whole number from a to b, including both
jersey = random.randint(1, 55)
print("Random jersey number:", jersey)

# math.sqrt returns the square root, so a 36 square foot court is this wide
side = math.sqrt(36)
print("Square court side:", side, "feet")

# math.ceil rounds up to the next whole number
average = sum(points_per_game) / len(points_per_game)
rounded_up = math.ceil(average)
print("Points per game:", average, "-> rounded up:", rounded_up)
```

```python
# CODE_RUNNER:

#!pip install flask

from flask import Flask, jsonify

app = Flask(__name__)

# A route is a procedure with a web address attached to it
@app.route("/score")
def score():
    return jsonify({"home": 58, "away": 52, "quarter": 3})

@app.route("/roster")
def roster():
    return jsonify({"players": ["Ryden", "Aarav", "Sam", "Bob", "Jill"]})

@app.route("/games")
def games():
    return jsonify({"games": ["2026-11-07", "2026-11-14", "2026-11-21", "2026-11-28"]});

# test_client visits the addresses without needing a browser
client = app.test_client()

print("/score says:", client.get("/score").get_json())
print("/roster says:", client.get("/roster").get_json())
print("/games says:", client.get("/games").get_json())
```

3 Ways to import

```python
# CODE_RUNNER:

import math                    # import the whole library
from random import randint     # import one procedure out of a library
import statistics as s     # import the library under a shorter nickname

# The whole library was imported, so the math. prefix is required
print("Hoop circumference:", round(2 * math.pi * 0.75, 2), "feet")

# randint was imported by name, so it is called with no prefix
print("Coin toss, 1 means home ball:", randint(1, 2))

# stats is a nickname that stands in for statistics
print("Points per game:", s.mean([24, 18, 31]))
```

```python
import math
import random
import datetime
import fractions

import json

gamedata = [
    {"date": "2026-11-07", "home": random.randint(0, 100), "away": random.randint(0, 100), "opponent": "jayden"},
    {"date": "2026-11-14", "home": random.randint(0, 100), "away": random.randint(0, 100), "opponent": "guy"},
    {"date": "2026-11-21", "home": random.randint(0, 100), "away": random.randint(0, 100), "opponent": "jill"},
    {"date": "2026-11-28", "home": random.randint(0, 100), "away": random.randint(0, 100), "opponent": "bob"}
]

# Summarize season
def summarize_season(games):
    """
    summarizes the given season
    
    Parameters:
        games (list): a list of dictionaries, each representing a game with keys "date" (string), "home" (integer), "away" (integer), and "opponent" (string)

    Returns:
        dict: a python dictionary containing the total points, average points, best game score, and points by opponent
    """
    
    report = {
        "total points": 0,
        "average points": 0,
        "best game": None,
        "points by opponent": {}
    }
    
    points = []
    
    for i in range(len(games)):
        points.append(games[i]["home"] + games[i]["away"])
        
        opponent = games[i]["opponent"] # Find opponent name in the games table
        points_by_opponent = report["points by opponent"] # Find entry in the report table
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + games[i]["away"] # Add the away points to the report entry for that opponent
    
    report["total points"] = sum(points)
    report["average points"] = report["total points"] / len(games)
    report["best game"] = max(points)
    
    return report

the_report = summarize_season(gamedata)

game_that_is_next = datetime.datetime(2027, 1, 1)

the_report["days_until_next_game"] = (game_that_is_next - datetime.datetime.today()).days

print("Season summary:", json.dumps(the_report, indent=4))

print("average points as a fraction:", fractions.Fraction(the_report["average points"]).limit_denominator(1000))

# requirements.txt lists these libraries because they are not of the standard library, so they must be installed with pip install -r requirements.txt.
# json and datetime are part of the standard library, so they do not need to be listed in requirements.txt.
requirements = """pandas
Flask"""

print("========= requirements.txt =========")
print(requirements)
print("====================================")
```
