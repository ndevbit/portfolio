### Popcorn Hack #1

```python
is_weekday = False
homework_finished = True

if is_weekday: # Outer branch
    if homework_finished: # Inner branch
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
```

|is_weekday|homework_finished|Output|
|----------|-----------------|------|
|False     |False            |Weekend plan|
|True      |False             |Finish homework|
|False     |True             |Weekend plan|
|True      |True             |Free Time|

### Popcorn Hack #2

```python
logged_in = True
has_permission = True

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
```

### The code isn't nested because the if/else statement on line 6 is not indented

|---------|--------------|------|
