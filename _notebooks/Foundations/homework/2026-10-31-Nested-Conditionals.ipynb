### Homework

```python

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": False,
    "practice_today": False
}

# Write your complete nested conditional below.
# Do not delete the starter data. Test several value combinations.

logged_in = student["logged_in"]
homework_finished = student["homework_finished"]
lunch_packed = student["lunch_packed"]
practice_today = student["practice_today"]

if logged_in:
    if practice_today:
        if lunch_packed:
            print("Go to practice")
        else:
            print("Pack lunch first, then go to practice")
    else:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
else:
    print("Log in first")
```

|logged_in|homework_finished|lunch_packed|practice_today|output|
|---------|-----------------|------------|--------------|------|
|False    |True             |True        |False         |Log in first|
|True     |True             |True        |False         |Free time|
|True     |True             |True        |True          |Go to practice|
|True     |True             |False       |True          |Pack lunch first, then go to practice|
|True     |False            |False       |False         |Finish homework

The nested if statements are reliant on the out if statements because if the outer if statments do not go to the inner if statement, than the inner if statement does not run. Thus the inner if statements are dependent on the outer if statements' decision. For example, an inner if statement will not execute unless the out if statement it is in resolves to true
