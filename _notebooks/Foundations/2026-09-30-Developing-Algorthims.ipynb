Popcorn Hack 1

```python

# Step 1: make the hours list
hours = [3, 5, 2, 4] # Input list

# Step 2: start total at 0
total = 0 # Initialize "total" at 0

# Step 3: FOR EACH -> for ... in ...: add each h to total
for h in hours: # Iterate over elements, adding each element to the "total" variable
    total += h
    
# Step 4: LENGTH(hours) -> len(hours): compute the average
average = total / len(hours) # Compute the average

# Step 5: print total, then average
print(total) # Print variables
print(average)
```

Homework

```python
items    = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]      # in the closet now
target   = [30, 20, 25, 15, 30]    # what staff want on hand

# Step 1: COLLECT items below target, and how many of each are needed
short_items = []
short_amounts = []
for i in range(len(items)):
    # TODO 1: if in_stock[i] < target[i], append items[i] and target[i] - in_stock[i]
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])
    pass

# Step 2: SUM the shortages with a loop (no sum())
total_needed = 0
# TODO 2
for i in range(len(short_amounts)):
    total_needed += short_amounts[i]

# Step 3: FIND THE MAX shortage (track its position so you know the item's name)
# TODO 3
max_shortage_index = 0
for i in range(len(short_amounts)):
    if short_amounts[i] > short_amounts[max_shortage_index]:
        max_shortage_index = i 

# Step 4: COMBINE: print "Fully stocked!" if nothing is short, otherwise the report
# TODO 4
# print("Replace this line")

if len(short_items) == 0: # No shortages
    print("Fully stocked!")
else: # Shortage(s) found
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
        
    print(f"Total items needed {total_needed}")
    print(f"Most urgent: {short_items[max_shortage_index]} (short by {short_amounts[max_shortage_index]})")
```
