```python
def isBoulevard(streetName):
    target = "boulevard"
    for i in range(len(streetName)):
        if streetName[i:i+len(target)].lower() == target:
            return True
    return False

street = input("Enter street name: ")

if isBoulevard(street):
    print(street, "is a boulevard")
else:
    print(street, "is not a boulevard")
```

### SEARCH ALGORITHM
In python

```python
streetNames = ["Sigma drive", "Shape drive", "Car drive"]

def gpsFindStreet(streetName, start):
    for i in range(start, len(streetNames)):
        if streetNames[i] == streetName:
            return i

    print("GPS ERROR: COULD NOT FIND STREET:", streetName)

    return -1

startIndex = 1

targetStreet = input("Find street")

foundStreet = gpsFindStreet(targetStreet, startIndex)

if foundStreet != -1:
    print("Found \"", streetNames[foundStreet], "\"")
else:
    print("No street here named \"", streetNames[foundStreet], "\" past index", str(startIndex))
```
